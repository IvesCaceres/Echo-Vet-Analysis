# Aggregate results viewer
Defaults to invented SYNTHETIC_DEMO examples. REAL_AGGREGATED reads only exact hash-reviewed public aggregate CSVs, never a clinical database. Local content approval does not authorize remote publication.


```
from pathlib import Path
import os, sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts' / 'editorial_release.py').is_file() and (p / 'published' / 'editorial-manifest.json').is_file())
sys.path.insert(0, str(ROOT / 'scripts'))
from editorial_release import load_editorial_tables
from aggregate_release import synthetic_example
from display_helpers import display
import pandas as pd
import matplotlib.pyplot as plt
```

```
kind = os.environ.get('ECHO_RESULTS_KIND', 'SYNTHETIC_DEMO')
if kind == 'SYNTHETIC_DEMO':
    example = synthetic_example()['tables'][0]
    tables = {'species': [{'category_en': r['category_code'], 'count': r['count'], 'denominator_records': example['denominator'], 'percentage': r['percentage']} for r in example['rows']]}
elif kind == 'REAL_AGGREGATED':
    tables = load_editorial_tables(ROOT)
else:
    raise ValueError('Choose SYNTHETIC_DEMO or REAL_AGGREGATED explicitly')
print(kind)
for name, rows in tables.items():
    print(name)
    display(pd.DataFrame(rows))
```

```
for name in ['species', 'sex']:
    if name not in tables:
        continue
    rows = tables[name]
    fig, axis = plt.subplots(figsize=(9, 4))
    axis.barh([r['category_en'] for r in rows], [int(r['count']) for r in rows])
    axis.set(title=kind + ' | ' + name, xlabel='Report records; denominator ' + str(rows[0]['denominator_records']))
    axis.invert_yaxis()
    fig.tight_layout()
    plt.show()
    plt.close(fig)
```
